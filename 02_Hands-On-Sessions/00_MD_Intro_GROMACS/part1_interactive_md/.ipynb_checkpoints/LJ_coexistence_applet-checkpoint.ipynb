## Launch the applet
Run the next cell. Choose a **Preset** → **Reset**.

- Use **Play ▶** / **Pause** to run or stop continuously
- Or **Step once** to advance a single frame (always refreshes the plots)
- **Record T, P, ρ, g(r)** stores a snapshot of the status values when the system looks steady

Three panels: **atoms** | **T & P vs time** | **g(r)**.

**Atom colours:** colour shows **speed**, not atom type (plasma scale: darker ≈ slower, brighter/yellow ≈ faster). Hotter / faster particles stand out after you change T* or when the run is unstable.

**Status line:** N, L, number density **ρ = N/L²**, kinetic **T**, wall **P**, Δt.

If an old frozen run is still active: **Interrupt** the kernel, then re-run the import + `show_applet()` cells.

**How to read phase:** gas ≈ sparse + flat g(r); liquid ≈ dense disordered + decaying g(r) shells; solid ≈ lattice + sharp lasting g(r) peaks. Interface presets show two phases at once.

## Exercises (work together with the room)

### 1. Timestep
Preset **Liquid**. Run at Δt ≈ **0.015** (stable), then set Δt to **~0.08–0.12** until you see **SIMULATION BLEW UP**. Reset and restore Δt ≈ 0.015.

1. What goes wrong when Δt is too large?
2. What about the potential sets a safe timestep?
3. Why is increasing `dt` a bad way to “speed up” a GROMACS run?
4. When it blows up, what happens to **T** and **P** on the middle panel (and in the status line)?

### 2. Phases + g(r) (+ density)
Try presets **Gas**, **Liquid**, **Solid**. For each: Reset → run briefly → **Record**.

1. How does g(r) distinguish liquid from solid when both look “dense”?
2. How do you know the system is **equilibrated**?
3. Compare your recorded **ρ** for gas vs liquid vs solid. Which is highest, and why?
4. Is recorded **P** noisier than **T**? Why might that be (small system, walls)?

### 3. Interfaces
Try **Liquid + gas**, **Solid + gas**, and **Solid + liquid**.

1. Run at the default T* for each preset.
2. Change T* a little (±0.1), Reset, run again — does the interface grow, shrink, or melt?
3. **Record** when it looks steady (logs T, P, ρ, max g(r)).
4. For **Liquid + gas**, is the recorded **ρ** closer to liquid or gas? Why is a single box-average density tricky for a two-phase system?

### 4. Discussion — validating a force field
Connect to Sofia’s lecture (*validation and predictive realism*):

1. This toy model is pure **Lennard-Jones** (natural units here). Which observables did you just use that could test whether the interaction is “right”? (Include T, P, ρ, g(r), phases.)
2. Molecular force fields (OPLS, ClayFF, …) also have **Coulomb** (partial charges). You did not simulate charges today — but how would validation work *in the same spirit*?
3. Why is matching a single snapshot a weak test compared with clear phases + g(r) + (later) a coexistence curve of densities?
